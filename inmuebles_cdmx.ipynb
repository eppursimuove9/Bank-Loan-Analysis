# ¿Cuáles son los valores promedio de alquiler por tipo de propiedad?

# ¿Cuál es el porcentaje de cada tipo de propiedad en nuestra base de datos?

```python
datos.Tipo.unique()
```

```python
inmuebles_comerciales = ['Cocineta', 'Casa', 'Conjunto Comercial/Sala', 'Departamento',
       'Casa de Condominio', 'Edificio Completo', 'Departamento en Hotel',
       'Tienda/Salón', 'Casa Comercial', 'Casa de Vecindad',
       'Terreno Estándar', 'Cochera/Estacionamiento', 'Loft',
       'Galpón/Depósito/Almacén', 'Tienda en Centro Comercial', 'Rancho',
       'Posada/Chalé', 'Estudio', 'Hotel', 'Loteo/Condominio',
       'Industria']
```

```python
df = datos.query('@inmuebles_comerciales not in Tipo')
datos.head()
```

```python
def_tipo_precio = df.groupby('Tipo')[['Valor']].mean().sort_values('Valor')
def_tipo_precio.plot(kind='barh', figsize=(12,8), color= 'purple')
```

```python
df.Tipo.unique()
```

```python
df.Tipo.value_counts(normalize=True)
```

```python
df.query('Tipo == "Departamento"')
```

```python
df = df.query('Tipo == "Departamento"')
df.head()
```

```python
porcentaje_tipo = df.Tipo.value_counts(normalize=True)
porcentaje_tipo.plot(kind='barh', figsize=(12,8), color= 'green',xlabel='Tipos',ylabel='Porcentaje')
```
